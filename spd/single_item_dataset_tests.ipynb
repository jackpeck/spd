```python
def make_single_item_dataset_dict(dataset_dict):
    ds = dataset_dict["train"].select([0])
    return DatasetDict({key: ds for key in dataset_dict.keys()})
```

```python
class SingleItemDataset(Dataset):
    def __init__(self, length, item):
        self.len = length
        self.item = item

    def __len__(self):
        return self.length

    def __getitem__(self, idx):
        return self.item
```

```python
ds = SingleItemDataset(len(dataset), dataset[0])
```
